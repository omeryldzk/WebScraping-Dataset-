```python
import pandas as pd

# Load the data and specify the 'scholarshipRate' and 'idOSYM' columns as strings
csvData4 = pd.read_csv("updated_2019-2024_with_filled_counts.csv", encoding="utf-8")
df = pd.DataFrame(csvData4)

# Ensure the year column is available and filter by it
df_2019 = df[df['academicYear'] == 2019]
df_2020 = df[df['academicYear'] == 2020]
df_2021 = df[df['academicYear'] == 2021]
df_2022 = df[df['academicYear'] == 2022]
df_2023 = df[df['academicYear'] == 2023]
df_2024 = df[df['academicYear'] == 2024]
# Find common unique identifiers
common_ids = set(df_2019['idOSYM']).intersection(set(df_2020['idOSYM'])).intersection(set(df_2021['idOSYM'])).intersection(set(df_2022['idOSYM'])).intersection(set(df_2023['idOSYM'])).intersection(set(df_2024['idOSYM']))
print("Number of 2019 IDs:", df_2019.shape)
print("Number of 2020 IDs:", df_2020.shape)
print("Number of common IDs:", len(common_ids))
# Filter the DataFrame based on common IDs
df_2019_x = df_2019[df_2019['idOSYM'].isin(common_ids)]
df_2020_y = df_2020[df_2020['idOSYM'].isin(common_ids)]
df_2019_x.drop_duplicates(subset='idOSYM', keep='first', inplace=True)
df_2020_y.drop_duplicates(subset='idOSYM', keep='first', inplace=True)

print("Number of 2019 IDs x:", df_2019_x.shape)
print("Number of 2020 IDs y:", df_2020_y.shape)
# Drop duplicates based on 'idOSYM'


with open("test2.csv", "w", newline="", encoding="utf-8") as aFile:
    df_2020_y.to_csv(aFile, index=False)
```
